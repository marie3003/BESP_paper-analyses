```python
from Bio import Phylo

trees_path = "../results/run1/simulated_data/linearconstant/expgrowthfast/expgrowthfast.trees"
out_path = "../results/run1/simulated_data/linearconstant/expgrowthfast/expgrowthfast_1_dates.tsv"

mu = 4.6e-8  # substitutions per site per time unit
alignment_length = 10000000  # total number of sites in the alignment

# Read only the first tree from the .trees file
tree_iter = Phylo.parse(trees_path, "newick")
next(tree_iter)      # skip first tree
tree = next(tree_iter)  # use second tree

with open(out_path, "w") as f:
    f.write("name\tdate\n")

    for tip in tree.get_terminals():
        root_to_tip_dist = tree.distance(tree.root, tip)
        #sampling_time = root_to_tip_dist / (mu * alignment_length)

        f.write(f"{tip.name}\t{root_to_tip_dist}\n")

print(f"Wrote dates to {out_path}")
```

```python
import numpy as np

nsamples = 250
times = np.arange(0, 101)  # interval boundaries: 0..60
N0 = 2000.0
r = 0.01

areas = (N0 / r) * (np.exp(-r * times[:-1]) - np.exp(-r * times[1:]))

expected_counts = nsamples * areas / areas.sum()
rates = expected_counts / np.diff(times)

# Add zero rate after the final boundary
rates = np.append(rates, 0.0)

rate_string = " ".join(f"{x:.8f}" for x in rates)
change_times_string = " ".join(str(t) for t in times[1:])

print(f'<reaction spec="Reaction" rate="{rate_string}" changeTimes="{change_times_string}">0 -> pop</reaction>')
```

```python
import numpy as np

nsamples = 250

N = 1000.0
N_bottleneck = 100.0

# Bottleneck settings
bottleneck_start = 20.0
bottleneck_end = 23.0
bottleneck_depth = 100.0

# Sampling window
sampling_start = 0.0
sampling_end = 30.0

# Interval lengths
dt_before = bottleneck_start - sampling_start
dt_bottleneck = bottleneck_end - bottleneck_start
dt_after = sampling_end - bottleneck_end

# Areas under N(t)
area_before = N * dt_before
area_bottleneck = N_bottleneck * dt_bottleneck
area_after = N * dt_after

areas = np.array([area_before, area_bottleneck, area_after])
dts = np.array([dt_before, dt_bottleneck, dt_after])

# Expected sample counts per interval, proportional to population-size area
expected_counts = nsamples * areas / areas.sum()

# Convert expected counts to per-year rates
rates = expected_counts / dts

# Add final zero rate after sampling_end
rates = np.append(rates, 0.0)

rate_string = " ".join(f"{x:.8f}" for x in rates)

# Change times: start bottleneck, end bottleneck, end sampling
change_times = [bottleneck_start, bottleneck_end, sampling_end]
change_times_string = " ".join(f"{x:g}" for x in change_times)

print(
    f'<reaction spec="Reaction" rate="{rate_string}" '
    f'changeTimes="{change_times_string}">0 -> pop</reaction>'
)
```
